```python
#Q1
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

temperature = temperature + 2

fahrenheit = (9/5) * temperature + 32

above_32 = temperature[temperature > 32]

count = np.sum(temperature > 32)

print("Corrected temperature:", temperature)
print("Temperature in Fahrenheit:", fahrenheit)
print("Readings greater than 32°C:", above_32)
print("Number of readings exceeding 32°C:", count)
```

```python
#2
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Total steps:", np.sum(steps))
print("Mean steps:", np.mean(steps))
print("Maximum:", np.max(steps))
print("Minimum:", np.min(steps))
print("Total steps for each day:", np.sum(steps, axis=0))
print("Total steps for each user:", np.sum(steps, axis=1))

max_position = np.unravel_index(np.argmax(steps), steps.shape)
print("Position of maximum:", max_position)
```

```python
#Q3
import numpy as np

original = np.array([1, 2, 3, 4, 5, 6])

subset = original[1:5]
subset[0] = 999

print("Original:", original)
print("Subset:", subset)

copied_array = original[1:5].copy()
copied_array[0] = 500

print("Original:", original)
print("Copied array:", copied_array)

arr = np.arange(1, 13).reshape(3, 4)

print("3 x 4 matrix:")
print(arr)

print("First row:", arr[0])
print("Last row:", arr[-1])
print("Second column:", arr[:, 1])
print("Rows 1-2 and columns 2-3:")
print(arr[0:2, 1:3])

flattened = arr.flatten()
raveled = arr.ravel()

print("Flatten:", flattened)
print("Ravel:", raveled)

raveled[0] = 100
print("After modifying ravel:")
print("Original:", arr)
print("Ravel:", raveled)

flattened[1] = 200
print("After modifying flatten:")
print("Original:", arr)
print("Flatten:", flattened)

print("Shape:", arr.shape)
print("Dimensions:", arr.ndim)
print("Size:", arr.size)
print("Data type:", arr.dtype)
```

```python
#Q4
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("Shape:", X.shape)
print("Dimensions:", X.ndim)

print("X.T:")
print(X.T)

XT_X = X.T @ X
print("X.T @ X:")
print(XT_X)

inverse = np.linalg.inv(XT_X)
print("Inverse:")
print(inverse)

beta = inverse @ X.T @ y
print("Coefficients:", beta)

new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta

print("Predicted assistance score:", predicted_score)
```
